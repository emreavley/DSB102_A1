```python
import pandas as pd
import statsmodels.api as sm
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import mean_squared_error
import warnings; warnings.filterwarnings('ignore')

df = pd.read_csv("retail_dynamics_weekly_sales.csv")
y = df[["StoreID","Week", "Sales"]]

# print(df.head())

# Next is to make a function to standardise "Display", I want to convert a No to 0 and a yes to 1
def StandardisingDisplay(row):
    stringToBeConverted = row["Display"]
    if stringToBeConverted[0] == "N" or stringToBeConverted[0] == "n":
        return 0
    else:
        return 1

df["Display"] = df.apply(StandardisingDisplay, axis=1)

y_train = y[y["Week"] <= 8]
y_test = y[y["Week"] > 8]
y_train = y_train.drop(columns="Week")
y_test = y_test.drop(columns="Week")
```

```python
mean_sales_by_store_train = y_train.groupby("StoreID")["Sales"].mean()
mean_sales_by_store_test = y_test.groupby("StoreID")["Sales"].mean()

comparison_df = pd.DataFrame({"train_mean": mean_sales_by_store_train, "test_mean": mean_sales_by_store_test}) 

mse = ((comparison_df["test_mean"] - comparison_df["train_mean"]) ** 2).mean()
rmse = np.sqrt(mse)

print(f"RMSE: {round(rmse, 2)}")
```
