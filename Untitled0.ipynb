3. Create and train the model:
  - n_estimators indicates the number of trees; less trees = unstable predictions, more trees = longer training times + less significant improvements in predictions over time
n_estimators = 100 to balance predictive stability and effiency

  - max_depth indicates how deep the trees can grow; low max depth means underfitting, could miss important info. large max depth (or max_depth = none) could fit training data too closely meaning overfitting

  - random_state controls the reproducibility of the training process; from my understanding it doesn't matter what number this is as long as it's consistent.
  
4. Predict the held out weeks and compare it with the actual data to measure the errors
5. Evaluation, using MAE and RSME

```python
import pandas as pd
import numpy as np

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error

# Load the data
data = pd.read_csv("retail_dynamics_weekly_sales.csv")
# Cleaned data?? loaded as a csv file or something else?

# Allocate features to X and select the target (sales)
X = data[["Week", "Price", "Display", "Media"]]
Y = data["Sales"]

# Split the training and test data up
X_train = X.iloc[:8]
X_test = X.iloc[8:]

Y_train = Y.iloc[:8]
Y_test = Y.iloc[8:]

# Create the model
model = RandomForestRegressor(n_estimators = 100, max_depth = 3, random_state = 5)

# Train the model on the first 8 weeks
model.fit(X_train, Y_train)

# Predict the sales for weeks 9 and 10
predictions = model.predict(X_test)

print("Actual sales:", Y_test.tolist()) # convert data into a list
print("Predicted sales:", predictions.round(2)) # round predicted sales

# Calculate evaluation metrics
MAE = mean_absolute_error(Y_test, predictions)
RSME = np.sqrt(mean_squared_error(Y_test, predictions))

# Feature importance??
```
